```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("../data/StudentsPerformance.csv")
```

```python
df
```

```python
categorical_cols = df.select_dtypes(include="str").columns
categorical_cols
```

```python
for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].unique())
```

```python
df_encoded = pd.get_dummies( df, 
   columns=categorical_cols, 
   dtype=int
)
```

```python
df_encoded
```

```python
df_encoded.shape
```

```python
df_encoded.columns
```

```python
df_encoded.dtypes
```

```python
numeric_cols = df.select_dtypes(include="number").columns
numeric_cols
```

```python
scaler = StandardScaler()
```

```python
df_encoded[numeric_cols] = scaler.fit_transform(df_encoded[numeric_cols])
```

```python
df_encoded[numeric_cols].head()
```

```python
df_encoded[numeric_cols].mean()
```

```python
df_encoded[numeric_cols].std()
```

```python
df_encoded["average_score"] = df[
    ["math score", "reading score", "writing score"]
].mean(axis=1)
```

```python
df_encoded[[
    "math score",
    "reading score",
    "writing score",
    "average_score"
]]
```

La variable average_score représente la performance académique globale de chaque élève en calculant la moyenne de ses scores en mathématiques, lecture et écriture. Elle permet de synthétiser les trois performances individuelles en un indicateur unique, tout en restant facilement interprétable sur une échelle de 0 à 100.
